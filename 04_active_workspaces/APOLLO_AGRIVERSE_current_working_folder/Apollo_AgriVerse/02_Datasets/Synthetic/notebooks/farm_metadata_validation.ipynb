```python
import pandas as pd
```

```python
df = pd.read_csv("../Raw/farm_metadata.csv")
df.head()
```

```python
df.info()
```

```python
print("Latitude :", df["latitude"].min(), "-", df["latitude"].max())
print("Longitude:", df["longitude"].min(), "-", df["longitude"].max())
print("Area      :", df["area_acres"].min(), "-", df["area_acres"].max())
```
