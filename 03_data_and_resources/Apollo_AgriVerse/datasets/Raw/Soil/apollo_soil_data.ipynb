```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)
num_rows = 5000  

# 1. Fixed 'H' to lowercase 'h' to remove warning
timestamps = pd.date_range(start='2026-01-01', periods=num_rows, freq='h')

regions = ['Maharashtra', 'Punjab', 'Karnataka', 'Uttar Pradesh', 'Gujarat', 'Tamil Nadu']
soil_types = ['Alluvial', 'Black', 'Red', 'Laterite', 'Arid']

region_col = np.random.choice(regions, num_rows)
soil_type_col = np.random.choice(soil_types, num_rows)

soil_ph = np.round(np.random.uniform(4.5, 8.5, num_rows), 2)
nitrogen = np.round(np.random.uniform(20.0, 140.0, num_rows), 2)
phosphorus = np.round(np.random.uniform(10.0, 80.0, num_rows), 2)
potassium = np.round(np.random.uniform(15.0, 120.0, num_rows), 2)

# Adjusted moisture range slightly so critical dry states happen more frequently
soil_moisture = np.round(np.random.uniform(8.0, 45.0, num_rows), 2)
soil_temp = np.round(np.random.uniform(18.0, 42.0, num_rows), 2)

dryness_status = []
hydrogel_trigger = []

for moisture in soil_moisture:
    if moisture < 15.0:
        dryness_status.append('Critical_Dry')
        hydrogel_trigger.append(1)  # Trigger water release
    elif moisture < 23.0:
        dryness_status.append('Moderate_Dry')
        hydrogel_trigger.append(0)
    else:
        dryness_status.append('Normal')
        hydrogel_trigger.append(0)

df_soil_twin = pd.DataFrame({
    'Timestamp': timestamps,
    'Region_State': region_col,
    'Soil_Type': soil_type_col,
    'Soil_PH': soil_ph,
    'Nitrogen_N': nitrogen,
    'Phosphorus_P': phosphorus,
    'Potassium_K': potassium,
    'Soil_Moisture_Percent': soil_moisture,
    'Soil_Temperature_C': soil_temp,
    'Dryness_Status': dryness_status,
    'Hydrogel_Water_Release_Trigger': hydrogel_trigger
})

# Show rows where the hydrogel trigger activated (value is 1)
print("Sample rows where Hydrogel Trigger activated (Value = 1):")
display(df_soil_twin[df_soil_twin['Hydrogel_Water_Release_Trigger'] == 1].head(3))

print("\nTotal Dataset Shape:", df_soil_twin.shape)
```

```python
# Save strictly as a .csv file
output_csv_path = 'apollo_indian_soil_digital_twin.csv'
df_soil_twin.to_csv(output_csv_path, index=False)

print("Saved cleanly as a CSV file!")
```
