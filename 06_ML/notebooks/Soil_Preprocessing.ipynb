# Splitting The Dataset 

```python
# Split the single column into multiple columns based on commas
df_split = df.iloc[:, 0].str.split(',', expand=True)

# Assign the proper column headers
header_list = [
    'Timestamp', 'Region_State', 'Soil_Type', 'Soil_PH', 
    'Nitrogen_N', 'Phosphorus_P', 'Potassium_K', 
    'Soil_Moisture_Percent', 'Soil_Temperature_C', 
    'Dryness_Status', 'Hydrogel_Water_Release_Trigger'
]
df_split.columns = header_list

# Replace old df with our perfectly split dataframe
df = df_split

# Verify the new shape and columns
print("New Shape (Rows, Columns):", df.shape)
print(df.head(2))
```

# Data Preprocessing & Cleaning.

```python
# 1. Convert numeric columns from strings to actual numbers
numeric_cols = ['Soil_PH', 'Nitrogen_N', 'Phosphorus_P', 'Potassium_K', 
                'Soil_Moisture_Percent', 'Soil_Temperature_C', 'Hydrogel_Water_Release_Trigger']

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# 2. Check for any missing values after conversion
print("Missing values per column:\n", df.isnull().sum())

# 3. Fill or drop any missing values if found
df = df.dropna()

# 4. Save the final cleaned and preprocessed dataset to your Processed folder
output_path = r'D:\MindforgeAI Internship 2026 - 2026IT002- Sunaina Gaikwad\Apollo_AgriVerse\02_Datasets\Processed\cleaned_soil_dataset.csv'
df.to_csv(output_path, index=False)

print("\nSuccess! Cleaned and preprocessed dataset saved to Processed folder.")
print("Final Shape:", df.shape)
```
