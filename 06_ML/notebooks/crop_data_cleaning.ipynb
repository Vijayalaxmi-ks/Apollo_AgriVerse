```python
# ==========================================
# Apollo AgriVerse - Crop Dataset Expansion
# ==========================================

import pandas as pd
import numpy as np

# -----------------------------
# Load Original Dataset
# -----------------------------
df = pd.read_csv(r"D:\Apollo_AgriVerse\02_Datasets\Raw\Crop\Crop_recommendation.csv")

print("Original Dataset Shape:", df.shape)

# -----------------------------
# Clean Existing Dataset
# -----------------------------
df.columns = df.columns.str.strip().str.lower()

df["label"] = (
    df["label"]
    .astype(str)
    .str.strip()
    .str.lower()
)

# Remove duplicates
df.drop_duplicates(inplace=True)

# Fill missing values
numeric_cols = ["n","p","k","temperature","humidity","ph","rainfall"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())

# -----------------------------
# New Crops to Add
# -----------------------------
new_crops = [
    "sugarcane",
    "potato",
    "onion",
    "tomato",
    "cabbage",
    "cauliflower",
    "brinjal",
    "okra",
    "garlic",
    "ginger",
    "turmeric",
    "soybean",
    "groundnut",
    "mustard",
    "sunflower",
    "wheat",
    "barley",
    "bajra",
    "jowar",
    "ragi",
    "pea",
    "capsicum",
    "cucumber",
    "pumpkin",
    "spinach",
    "fenugreek",
    "coriander",
    "chilli"
]

# -----------------------------
# Generate New Samples
# -----------------------------
generated_rows = []

samples_per_crop = 100

for crop in new_crops:

    for i in range(samples_per_crop):

        row = {

            "n": np.random.randint(20,140),

            "p": np.random.randint(15,90),

            "k": np.random.randint(15,120),

            "temperature": round(np.random.uniform(18,38),2),

            "humidity": round(np.random.uniform(45,95),2),

            "ph": round(np.random.uniform(5.5,7.8),2),

            "rainfall": round(np.random.uniform(40,350),2),

            "label": crop

        }

        generated_rows.append(row)

new_df = pd.DataFrame(generated_rows)

print("Generated Samples:", new_df.shape)

# -----------------------------
# Merge Datasets
# -----------------------------
final_df = pd.concat([df,new_df],ignore_index=True)

# -----------------------------
# Final Cleaning
# -----------------------------
final_df.drop_duplicates(inplace=True)

final_df["temperature"] = final_df["temperature"].clip(10,50)
final_df["humidity"] = final_df["humidity"].clip(20,100)
final_df["ph"] = final_df["ph"].clip(3,9)
final_df["rainfall"] = final_df["rainfall"].clip(0,500)

# -----------------------------
# Shuffle Dataset
# -----------------------------
final_df = final_df.sample(frac=1,random_state=42).reset_index(drop=True)

# -----------------------------
# Save Dataset
# -----------------------------
save_path = r"D:\Apollo_AgriVerse\02_Datasets\Raw\Crop\Crop_recommendation.csv"

final_df.to_csv(save_path,index=False)

print("\nDataset Saved Successfully")

print("\nFinal Shape:",final_df.shape)

print("\nTotal Crops:",final_df["label"].nunique())

print("\nCrop Names:\n")

print(sorted(final_df["label"].unique()))

print("\nCrop Count:\n")

print(final_df["label"].value_counts())
```
