```python
import os
import random
import numpy as np
from PIL import Image, ImageDraw, ImageFilter

base_dir = r"D:\MindforgeAI Internship 2026 - 2026IT002- Sunaina Gaikwad\Apollo_AgriVerse"

# Define directories
hydrogel_dir = os.path.join(base_dir, "02_Datasets", "Raw", "Hydrogel", "Image_Dataset")
mulching_dir = os.path.join(base_dir, "02_Datasets", "Raw", "Mulching", "Image_Dataset")

for cat in ['dry_state', 'swollen_state']:
    os.makedirs(os.path.join(hydrogel_dir, cat), exist_ok=True)
for cat in ['intact_film', 'degraded_film']:
    os.makedirs(os.path.join(mulching_dir, cat), exist_ok=True)

print("Directories ready. Generating robust synthetic agricultural textures...")

# 1. Advanced Hydrogel Dataset Generator (150 images per state)
def generate_robust_hydrogel():
    for state in ['dry_state', 'swollen_state']:
        folder = os.path.join(hydrogel_dir, state)
        for i in range(150):  # Generates 150 unique, varied images
            # Create a base soil texture with random brownish earth tones
            base_color = (random.randint(40, 70), random.randint(30, 50), random.randint(20, 40))
            img = Image.new('RGB', (224, 224), color=base_color)
            
            # Add random background noise to simulate real soil grains
            np_img = np.array(img)
            noise = np.random.normal(0, 15, np_img.shape).astype(np.uint8)
            np_img = np.clip(np_img.astype(int) + noise, 0, 255).astype(np.uint8)
            img = Image.fromarray(np_img)
            
            draw = ImageDraw.Draw(img)
            num_spheres = random.randint(30, 60)
            
            for _ in range(num_spheres):
                x = random.randint(10, 214)
                y = random.randint(10, 214)
                
                if state == 'dry_state':
                    # Small, shrunken, dull brown/grey polymer granules
                    radius = random.randint(2, 5)
                    color = (random.randint(80, 110), random.randint(70, 90), random.randint(60, 80))
                else:
                    # Large, swollen, translucent blue/clear hydrogel spheres absorbing water
                    radius = random.randint(8, 18)
                    color = (random.randint(0, 50), random.randint(120, 200), random.randint(220, 255))
                    
                draw.ellipse([x-radius, y-radius, x+radius, y+radius], fill=color)
            
            # Apply slight blur to blend with soil naturally
            img = img.filter(ImageFilter.GaussianBlur(radius=0.5))
            img.save(os.path.join(folder, f"hydrogel_{state}_{i+1}.jpg"))

# 2. Advanced Mulching Paper Dataset Generator (150 images per state)
def generate_robust_mulching():
    for state in ['intact_film', 'degraded_film']:
        folder = os.path.join(mulching_dir, state)
        for i in range(150):  # Generates 150 unique, varied images
            # Dark agricultural plastic film background
            img = Image.new('RGB', (224, 224), color=(25, 25, 25))
            draw = ImageDraw.Draw(img)
            
            if state == 'intact_film':
                # Clean, structured parallel stripe lines representing fresh plastic mulch sheets
                for line_x in range(0, 224, 25):
                    stripe_color = (random.randint(35, 50), random.randint(35, 50), random.randint(35, 50))
                    draw.rectangle([line_x, 0, line_x + random.randint(12, 18), 224], fill=stripe_color)
            else:
                # Degraded film: torn edges, faded patches, soil spots, and cracks
                for line_x in range(0, 224, 25):
                    draw.rectangle([line_x, 0, line_x + 15, 224], fill=(45, 40, 35))
                
                # Add random tears, holes, and dirt accumulation
                for _ in range(8):
                    hx = random.randint(10, 200)
                    hy = random.randint(10, 200)
                    hw = random.randint(15, 40)
                    hh = random.randint(10, 25)
                    draw.ellipse([hx, hy, hx+hw, hy+hh], fill=(60, 50, 40)) # Soil patch showing through tear
                    
                for _ in range(6):
                    rx = random.randint(10, 200)
                    ry = random.randint(10, 200)
                    draw.line([rx, ry, rx + random.randint(20, 50), ry + random.randint(10, 30)], fill=(80, 70, 60), width=3)

            img.save(os.path.join(folder, f"mulch_{state}_{i+1}.jpg"))

generate_robust_hydrogel()
generate_robust_mulching()
print("Success! 150 rich, varied variations generated for Hydrogel and Mulching datasets.")
```
