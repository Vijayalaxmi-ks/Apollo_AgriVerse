```python
import os
import random
from PIL import Image

base_dir = r"D:\MindforgeAI Internship 2026 - 2026IT002- Sunaina Gaikwad\Apollo_AgriVerse"

# Categories to process
categories = {
    "Hydrogel": ['dry_state', 'swollen_state'],
    "Mulching": ['intact_film', 'degraded_film']
}

split_ratio = 0.8  # 80% Train, 10% Val, 10% Test

for category_name, classes in categories.items():
    raw_dir = os.path.join(base_dir, "02_Datasets", "Raw", category_name, "Image_Dataset")
    processed_dir = os.path.join(base_dir, "02_Datasets", "Processed", category_name, "Image_Dataset")
    
    print(f"\nProcessing category: {category_name}")
    
    for cls in classes:
        class_source_path = os.path.join(raw_dir, cls)
        if not os.path.exists(class_source_path):
            print(f"  Warning: Path not found {class_source_path}")
            continue
            
        # Collect images
        images = [f for f in os.listdir(class_source_path) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
        random.shuffle(images)
        
        train_count = int(len(images) * split_ratio)
        val_count = int(len(images) * 0.1)
        
        train_imgs = images[:train_count]
        val_imgs = images[train_count:train_count + val_count]
        test_imgs = images[train_count + val_count:]
        
        # Helper function to save splits
        def save_split(img_list, split_name):
            split_class_dir = os.path.join(processed_dir, split_name, cls)
            os.makedirs(split_class_dir, exist_ok=True)
            
            saved_count = 0
            for img_name in img_list:
                try:
                    img_path = os.path.join(class_source_path, img_name)
                    img = Image.open(img_path).convert('RGB')
                    img_resized = img.resize((224, 224))
                    save_path = os.path.join(split_class_dir, img_name)
                    img_resized.save(save_path)
                    saved_count += 1
                except Exception as e:
                    pass
            return saved_count

        t_saved = save_split(train_imgs, 'train')
        v_saved = save_split(val_imgs, 'val')
        s_saved = save_split(test_imgs, 'test')
        print(f"  -> {cls}: Saved {t_saved} train, {v_saved} val, {s_saved} test images.")

print("\nAll Hydrogel and Mulching datasets successfully preprocessed and structured!")
```
