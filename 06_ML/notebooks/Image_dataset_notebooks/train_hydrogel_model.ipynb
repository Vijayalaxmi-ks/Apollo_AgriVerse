```python
import os
import tensorflow as tf
from tensorflow.keras.preprocessing import image_dataset_from_directory
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras import layers, models

# 1. Correct path pointing to Hydrogel/Image_Dataset
base_dir = r"D:\MindforgeAI Internship 2026 - 2026IT002- Sunaina Gaikwad\Apollo_AgriVerse"
hydrogel_dir = os.path.join(base_dir, "02_Datasets", "Processed", "Hydrogel", "Image_Dataset")

train_dir = os.path.join(hydrogel_dir, "train")
val_dir = os.path.join(hydrogel_dir, "val")

print(f"Loading Hydrogel/Mulch images from: {train_dir}")

# 2. Hyperparameters
IMG_SIZE = (224, 224)
BATCH_SIZE = 8
EPOCHS = 10
AUTOTUNE = tf.data.AUTOTUNE

# 3. Load Datasets
train_dataset = image_dataset_from_directory(
    train_dir,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=True
)

val_dataset = image_dataset_from_directory(
    val_dir,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)

class_names = train_dataset.class_names
print("\nDetected Hydrogel/Mulch Classes:", class_names)

# Optimize performance
train_dataset = train_dataset.cache().prefetch(buffer_size=AUTOTUNE)
val_dataset = val_dataset.cache().prefetch(buffer_size=AUTOTUNE)

# 4. Build Model safely
base_model = MobileNetV2(input_shape=(224, 224, 3),
                         include_top=False,
                         weights='imagenet')
base_model.trainable = False

model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.3),
    layers.Dense(128, activation='relu'),
    layers.Dense(len(class_names), activation='softmax')
])

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 5. Train the Model
print("\n--- Hydrogel Model Training Started ---")
history = model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=EPOCHS
)

# 6. Save Model Artifact
models_output_dir = os.path.join(base_dir, "06_ML", "models")
os.makedirs(models_output_dir, exist_ok=True)
model_save_path = os.path.join(models_output_dir, "hydrogel_model.keras")
model.save(model_save_path)
print(f"\nHydrogel Model successfully trained and saved at: {model_save_path}")
```
