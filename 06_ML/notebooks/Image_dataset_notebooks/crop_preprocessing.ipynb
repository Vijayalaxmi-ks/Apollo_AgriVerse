```python
import os
import random
from PIL import Image

base_dir = r"D:\MindforgeAI Internship 2026 - 2026IT002- Sunaina Gaikwad\Apollo_AgriVerse"
# Added 'Grapes' to match your exact File Explorer path shown in the screenshot
raw_crop_dir = os.path.join(base_dir, "03_Images", "Dataset", "Crop_Images", "Grapes")
processed_crop_dir = os.path.join(base_dir, "02_Datasets", "Processed", "Crop", "Image_Dataset")

crop_classes = ['01_Bud_Burst', '02_Vegetative', '03_Flowering', '04_Fruit_Development', '05_Harvest']
split_ratio = 0.8

for cls in crop_classes:
    class_path = os.path.join(raw_crop_dir, cls)
    if not os.path.exists(class_path):
        print(f"Skipping {cls}, folder not found at: {class_path}")
        continue
        
    images = [f for f in os.listdir(class_path) if f.lower().endswith(('.png', '.jpg', '.jpeg', '.avif'))]
    random.shuffle(images)
    
    train_count = int(len(images) * split_ratio)
    val_count = int(len(images) * 0.1)
    
    train_imgs = images[:train_count]
    val_imgs = images[train_count:train_count + val_count]
    test_imgs = images[train_count + val_count:]
    
    def save_split(img_list, split_name):
        split_dir = os.path.join(processed_crop_dir, split_name, cls)
        os.makedirs(split_dir, exist_ok=True)
        count = 0
        for img_name in img_list:
            try:
                img_path = os.path.join(class_path, img_name)
                img = Image.open(img_path).convert('RGB')
                img.resize((224, 224)).save(os.path.join(split_dir, f"{cls}_{img_name}.jpg"))
                count += 1
            except Exception as e:
                pass
        return count

    t = save_split(train_imgs, 'train')
    v = save_split(val_imgs, 'val')
    s = save_split(test_imgs, 'test')
    print(f"Processed {cls}: {t} train, {v} val, {s} test saved.")

print("\nCrop images successfully preprocessed and standardized!")
```
