```python
from PIL import Image

# Check for any broken or corrupted images in both folders
corrupted_count = 0
for dataset_type in ["Orignal-Dataset", "CyAUG-Dataset"]:
    type_path = os.path.join(raw_image_dir, dataset_type)
    for root, dirs, files in os.walk(type_path):
        for file in files:
            file_path = os.path.join(root, file)
            try:
                with Image.open(file_path) as img:
                    img.verify()
            except Exception as e:
                print(f"Found corrupted file: {file_path}")
                corrupted_count += 1

print(f"Image integrity check complete. Corrupted files found and noted: {corrupted_count}")
```

```python
# Define paths for the processed dataset split folders
processed_soil_dir = os.path.join(base_dir, "02_Datasets", "Processed", "Soil", "Image_Dataset")
soil_classes = ['Alluvial_Soil', 'Arid_Soil', 'Black_Soil', 'Laterite_Soil', 'Mountain_Soil', 'Red_Soil', 'Yellow_Soil']

# Create train, val, and test directories for every soil class
for split in ['train', 'val', 'test']:
    for soil in soil_classes:
        os.makedirs(os.path.join(processed_soil_dir, split, soil), exist_ok=True)

print("Processed train, validation, and test folders successfully created!")
```

```python
import os
import random
from PIL import Image

base_dir = r"D:\MindforgeAI Internship 2026 - 2026IT002- Sunaina Gaikwad\Apollo_AgriVerse"

# Let's dynamically find where 'Orignal-Dataset' or 'Soil' images are located
found_source_dir = None
for root, dirs, files in os.walk(base_dir):
    if "Orignal-Dataset" in dirs:
        found_source_dir = os.path.join(root, "Orignal-Dataset")
        break

print("Found source directory:", found_source_dir)

if found_source_dir:
    parent_raw_dir = os.path.dirname(found_source_dir)
    dest_dir = os.path.join(base_dir, "02_Datasets", "Processed", "Soil", "Image_Dataset")
    soil_classes = ['Alluvial_Soil', 'Arid_Soil', 'Black_Soil', 'Laterite_Soil', 'Mountain_Soil', 'Red_Soil', 'Yellow_Soil']
    
    for dataset_type in ["Orignal-Dataset", "CyAUG-Dataset"]:
        source_dir = os.path.join(parent_raw_dir, dataset_type)
        if not os.path.exists(source_dir):
            print(f"Skipping {dataset_type} (not found at {source_dir})")
            continue
            
        print(f"\nProcessing: {dataset_type}")
        for soil in soil_classes:
            class_path = os.path.join(source_dir, soil)
            if not os.path.exists(class_path):
                continue
                
            images = [f for f in os.listdir(class_path) if f.lower().endswith(('.png', '.jpg', '.jpeg', '.webp'))]
            if not images:
                continue
                
            random.shuffle(images)
            train_cut = int(len(images) * 0.8)
            val_cut = int(len(images) * 0.1) + train_cut
            
            for split_name, img_list in [('train', images[:train_cut]), ('val', images[train_cut:val_cut]), ('test', images[val_cut:])]:
                split_soil_dir = os.path.join(dest_dir, split_name, soil)
                os.makedirs(split_soil_dir, exist_ok=True)
                for img_name in img_list:
                    try:
                        img = Image.open(os.path.join(class_path, img_name)).convert('RGB')
                        img.resize((224, 224)).save(os.path.join(split_soil_dir, f"{dataset_type}_{img_name}"))
                    except Exception as e:
                        print(f"Error on {img_name}: {e}")
        print(f"Completed {dataset_type}!")
    print("\nProcessing complete! Check your processed folder.")
else:
    print("Could not locate 'Orignal-Dataset' folder automatically. Please check your folder names.")
```
