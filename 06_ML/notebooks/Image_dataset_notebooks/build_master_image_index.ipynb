```python
import os
import pandas as pd

base_dir = r"D:\MindforgeAI Internship 2026 - 2026IT002- Sunaina Gaikwad\Apollo_AgriVerse"
processed_root = os.path.join(base_dir, "02_Datasets", "Processed")
metadata_dir = os.path.join(base_dir, "02_Datasets", "Metadata")
os.makedirs(metadata_dir, exist_ok=True)

image_records = []

# Scan through all categories (Soil, Hydrogel, Mulching, Crop)
categories = os.listdir(processed_root)
for cat in categories:
    cat_path = os.path.join(processed_root, cat, "Image_Dataset")
    if not os.path.exists(cat_path):
        continue
        
    for split in ['train', 'val', 'test']:
        split_path = os.path.join(cat_path, split)
        if not os.path.exists(split_path):
            continue
            
        for class_name in os.listdir(split_path):
            class_path = os.path.join(split_path, class_name)
            if not os.path.isdir(class_path):
                continue
                
            for img_file in os.listdir(class_path):
                if img_file.lower().endswith(('.png', '.jpg', '.jpeg')):
                    relative_path = os.path.relpath(os.path.join(class_path, img_file), base_dir)
                    
                    image_records.append({
                        "image_id": f"IMG_{len(image_records)+1:04d}",
                        "category": cat,
                        "split": split,
                        "class_label": class_name,
                        "file_path": relative_path
                    })

# Convert to DataFrame and save as Master CSV
df_metadata = pd.DataFrame(image_records)
master_csv_path = os.path.join(metadata_dir, "Master_Image_Metadata.csv")
df_metadata.to_csv(master_csv_path, index=False)

print(f"Master Image Metadata successfully created with {len(df_metadata)} total records!")
print(f"Saved at: {master_csv_path}")
```
